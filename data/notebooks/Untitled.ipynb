```python
import pandas as pd
import numpy as np
df = pd.read_csv('/Users/spandandey/Desktop/courses/individual_project/Dashboard/Wine-value-optimization/data/raw/winemag-data-130k-v2.csv')

print(len(df))
# to remove unnecesary columns
if "Unnamed: 0" in df.columns:
    df=df.drop(columns=["Unnamed: 0"])

# to deal with duplicates
df=df.drop_duplicates()
print(len(df))

# fix data types
df["points"] = pd.to_numeric(df["points"], errors="coerce")
df["price"]  = pd.to_numeric(df["price"], errors="coerce")

print("Missing point:", df["points"].isna().sum())
print("Missing price:", df["price"].isna().sum())

# rows with proper prices
price_df= df[df["price"].notna()].copy()
price_df=price_df[price_df["price"]>0].copy()
print("Rows with valid price:", len(price_df))
```
